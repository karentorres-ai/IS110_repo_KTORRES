```python
def computepay(hours, rate):
    if hours <= 40:
        pay = hours * rate
    else:
        regular_pay = 40 * rate
        overtime_hours = hours - 40
        overtime_pay = overtime_hours * rate * 1.5
        pay = regular_pay + overtime_pay
    return pay

# Example usage:
hours_input = float(input('Enter Hours: '))
rate_input = float(input('Enter Rate: '))

calculated_pay = computepay(hours_input, rate_input)
print('Pay:', calculated_pay)
```

```python
def computegrade(score):
    try:
        score = float(score)
    except:
        return 'Bad score'

    if 97 <= score <= 100:
        return '1.0'
    elif 93 <= score < 97:
        return '1.25'
    elif 89 <= score < 93:
        return '1.5'
    elif 85 <= score < 89:
        return '1.75'
    elif 81 <= score < 85:
        return '2.0'
    elif 77 <= score < 81:
        return '2.25'
    elif 74 <= score < 77:
        return '2.5'
    elif 0 <= score < 74:
        return '3.0'
    else:
        return 'Bad score'
```

```python
while True:
    score_input = input('Enter a score (or "done" to finish): ')
    if score_input.lower() == 'done':
        break

    grade = computegrade(score_input)
    print(f'Score: {score_input}, Grade: {grade}')
```
