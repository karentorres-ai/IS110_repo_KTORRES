Exercise 2

```python
hours_str = input("Enter Hours: ")
rate_str = input("Enter Rate: ")

try:
    hours = float(hours_str)
    rate = float(rate_str)


    if hours > 40:
        regular_pay = 40 * rate
        overtime_hours = hours - 40
        overtime_pay = overtime_hours * (rate * 1.5)
        total_pay = regular_pay + overtime_pay
    else:
        total_pay = hours * rate

    print(f"Pay: {total_pay}")

except ValueError:
    print("Error: Please enter numeric values for hours and rate.")
```

CHALLENGE

```python
score_str = input("Enter score (0-100): ")

try:
    score = float(score_str)

    if 0 <= score <= 100:
        if score >= 97:
            grade = 1.0
        elif score >= 93:
            grade = 1.25
        elif score >= 89:
            grade = 1.5
        elif score >= 85:
            grade = 1.75
        elif score >= 81:
            grade = 2.0
        elif score >= 77:
            grade = 2.25
        elif score >= 74:
            grade = 2.5
        else:
            grade = 3.0
        print(f"Grade: {grade}")
    else:
        print("Error: Score is out of range (0-100).")
except ValueError:
    print("Error: Please enter a numeric value for the score.")
```
